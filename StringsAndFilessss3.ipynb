Cadenas de texto (Strings)
1. lower()
   Sintaxis: texto.lower()
Convierte caracteres alfabeticos de una cadena de textos a minusculas, es utilizado para organizar informacion y comparar textos sin importar
las mayusculas.

```python
# Convertir un texto a minúsculas
texto = "HOLA, ESTOY APRENDIENDO PYTHON"

resultado = texto.lower()

print("Texto original:", texto)
print("Texto en minúsculas:", resultado)
```

2. upper()
   Sintaxis: texto.upper()
Convierte todas las letras de una cadena de texto a mayusculas. utilizado para destacar informacion o presentar titulos.

```python
# Convertir un texto a mayúsculas
nombre = "antonio medina"

resultado = nombre.upper()

print("Nombre original:", nombre)
print("Nombre en mayúsculas:", resultado)
```

3. tittle()
    Sintaxis: texto.title()
Convierte la primera letra de cada palabra en mayuscula y las demas letras en minuscula.


```python
# Organizar el nombre de una persona
nombre = "luisa fernanda"

resultado = nombre.title()

print("Nombre organizado:", resultado)
```

4. capitalize()
Convierte la primera letra de una cadena en mayuscula y el resto del texto en minusculas.

```python
# Corregir la presentación de una frase
frase = "bienvenidos al curso de python"

resultado = frase.capitalize()

print("Frase corregida:", resultado)
```

5. strip()
Elimin los espacios en blanco que se encuentran al principio y al final de una cadena de texto. Utilizado para limpiar informacion ingresada
por los usuarios.

```python
# Eliminar espacios innecesarios
nombre = "   Antonio Medina   "

resultado = nombre.strip()

print("Texto original:", nombre)
print("Texto limpio:", resultado)
```

6. replace() 
    Sintaxis: texto.replace(texto_anterior, texto_nuevo)
Permite reemplazar una palabra o caracter por otro dentro de una cadena de texto.

```python
# Reemplazar una palabra dentro de una frase
frase = "Estoy aprendiendo Java"

resultado = frase.replace("Java", "Python")

print("Frase original:", frase)
print("Frase modificada:", resultado)
```

7. split()
divide una cadena de texto en diferentes elementos y los almacena dentro de una lista. Utiliza los espacios como separadores.

```python
# Dividir una frase en palabras
frase = "Python es un lenguaje de programación"

palabras = frase.split()

print("Frase original:", frase)
print("Palabras separadas:", palabras)
```

8. join()
    Sintaxis: separador.join(iterable)
Une varios elementos de una lista o coleccion en una sola cadena de texto.

```python
# Unir diferentes palabras en una frase
palabras = ["Ciencia", "de", "Datos"]

frase = " ".join(palabras)

print("Frase construida:", frase)
```

9. find()
    Sintaxis: texto.find(subcadena)
Busca una palabra o caracter dentro de una cadena y devuelve la posicion donde comienza su primera aparicion. Si no encuentra el elemento, devuelve 
-1.

```python
# Buscar la posición de una palabra
frase = "Estoy estudiando Python"

posicion = frase.find("Python")

print("Frase:", frase)
print("Posición de Python:", posicion)
```

10. count()
    Sintaxis: texto.count(subcadena)
Cuenta cuántas veces aparece una palabra, letra o carácter dentro de una cadena de texto.

```python
# Contar las apariciones de una letra
palabra = "programacion"

cantidad = palabra.count("a")

print("Palabra:", palabra)
print("Cantidad de letras a:", cantidad)
```

MANEJO DE ARCHIVOS (FILES)
11. open()
    Sintaxis: open(nombre_archivo, modo)
La función open() permite abrir archivos para leer, escribir o agregar información. También puede crear archivos nuevos dependiendo del modo
utilizado.

Los modos más utilizados son:
- r: lectura.
- w: escritura.
- a: agregar información.
- x: creación exclusiva de un archivo nuevo.

```python
# Crear un archivo de texto
archivo = open("estudiantes.txt", "w")

archivo.write("Lista de estudiantes del curso")

archivo.close()

print("Archivo creado correctamente.")
```

12. read()
permite leer el contenido de un archivo y devolverlo como una cadena de texto.

```python
# Leer el contenido de un archivo
archivo = open("estudiantes.txt", "r")

contenido = archivo.read()

print("Contenido del archivo:")
print(contenido)

archivo.close()
```

13. readline()
Lee una sola linea de un archivo cada vez que se ejecuta. Utilizada cuando se necesita consultar informacion por linea.

```python
# Leer la primera línea de un archivo
archivo = open("datos.txt", "w")

archivo.write("Nombre: Antonio\n")
archivo.write("Curso: Ciencia de Datos\n")

archivo.close()

archivo = open("datos.txt", "r")

primera_linea = archivo.readline()

print("Primera línea:", primera_linea)

archivo.close()
```

14. write()
Permite escribir informacion dentro de un archivo de texto. Recibe una cadena de caracteres y la guarda en el archivo abierto.

```python
# Escribir información en un archivo
archivo = open("registro.txt", "w")

archivo.write("Proyecto de Ciencia de Datos\n")
archivo.write("Actividad 3 de Python")

archivo.close()

print("Información guardada correctamente.")
```

15. close()
Cierra un archivo que se encuentre abierto, utilizado para liberar recursos y asegurar que los cambios se realizen correctamente.

```python
# Abrir y cerrar un archivo
archivo = open("registro.txt", "r")

print("Archivo abierto:", not archivo.closed)

archivo.close()

print("Archivo cerrado:", archivo.closed)
```

EJEMPLO CON ALGUNAS DE LAS FUNCIONES PRESENTADAS

```python
# Registro de estudiantes utilizando Python

estudiantes = [
    "antonio medina",
    "luisa fernanda",
    "alejo barrera"
]

# Organizar los nombres
nombres_organizados = []

for nombre in estudiantes:
    nombre = nombre.strip()
    nombre = nombre.title()
    nombres_organizados.append(nombre)

# Crear y escribir el archivo
archivo = open("registro_estudiantes.txt", "w")

for nombre in nombres_organizados:
    archivo.write(nombre + "\n")

archivo.close()

# Leer el archivo creado
archivo = open("registro_estudiantes.txt", "r")

contenido = archivo.read()

print("REGISTRO DE ESTUDIANTES")
print(contenido)

archivo.close()
```
