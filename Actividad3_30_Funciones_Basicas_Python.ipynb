# 30 funciones básicas de Python

**Actividad 3 – Introducción a Python**

Este cuaderno presenta treinta funciones incorporadas de Python. Cada sección explica para qué sirve una función y contiene un ejemplo práctico que se puede ejecutar.

**Nota:** algunas funciones, como `input()` y `help()`, interactúan con el usuario o muestran información de ayuda cuando se ejecutan.

## 1. Función `print()`

Muestra información en la pantalla. Se usa para presentar mensajes y resultados.

**Ejemplo práctico:**

```python
print("Hola mundo")
```

## 2. Función `len()`

Devuelve la cantidad de elementos de un objeto, como los caracteres de un texto o los elementos de una lista.

**Ejemplo práctico:**

```python
texto = "Python"
print(len(texto))
```

## 3. Función `type()`

Indica el tipo de dato de un valor o una variable.

**Ejemplo práctico:**

```python
numero = 25
print(type(numero))
```

## 4. Función `int()`

Convierte un valor compatible en un número entero.

**Ejemplo práctico:**

```python
numero = int("15")
print(numero)
```

## 5. Función `float()`

Convierte un valor compatible en un número decimal.

**Ejemplo práctico:**

```python
precio = float("12.5")
print(precio)
```

## 6. Función `str()`

Convierte un valor en una cadena de texto.

**Ejemplo práctico:**

```python
edad = 20
print("Edad: " + str(edad))
```

## 7. Función `bool()`

Convierte un valor a verdadero (`True`) o falso (`False`) según las reglas de Python.

**Ejemplo práctico:**

```python
valor = bool(1)
print(valor)
```

## 8. Función `abs()`

Devuelve el valor absoluto de un número, es decir, su distancia al cero sin signo negativo.

**Ejemplo práctico:**

```python
numero = -8
print(abs(numero))
```

## 9. Función `round()`

Redondea un número al entero más cercano o a la cantidad de decimales indicada.

**Ejemplo práctico:**

```python
precio = 3.14159
print(round(precio, 2))
```

## 10. Función `sum()`

Suma los elementos numéricos de un iterable, como una lista.

**Ejemplo práctico:**

```python
notas = [4, 3, 5]
print(sum(notas))
```

## 11. Función `max()`

Devuelve el elemento de mayor valor de un conjunto de elementos.

**Ejemplo práctico:**

```python
numeros = [8, 2, 15, 4]
print(max(numeros))
```

## 12. Función `min()`

Devuelve el elemento de menor valor de un conjunto de elementos.

**Ejemplo práctico:**

```python
numeros = [8, 2, 15, 4]
print(min(numeros))
```

## 13. Función `sorted()`

Devuelve una nueva lista con los elementos ordenados. Por defecto, los ordena de menor a mayor.

**Ejemplo práctico:**

```python
numeros = [5, 1, 3, 2]
print(sorted(numeros))
```

## 14. Función `range()`

Genera una secuencia de números enteros, muy útil para repetir instrucciones en un ciclo.

**Ejemplo práctico:**

```python
for numero in range(1, 6):
    print(numero)
```

## 15. Función `list()`

Crea una lista, que es una colección ordenada y modificable de elementos.

**Ejemplo práctico:**

```python
vocales = list("aeiou")
print(vocales)
```

## 16. Función `tuple()`

Crea una tupla, una colección ordenada que no se puede modificar después de crearla.

**Ejemplo práctico:**

```python
coordenadas = tuple([4, 7])
print(coordenadas)
```

## 17. Función `set()`

Crea un conjunto que conserva elementos únicos y elimina los duplicados.

**Ejemplo práctico:**

```python
numeros = set([1, 2, 2, 3])
print(numeros)
```

## 18. Función `dict()`

Crea un diccionario que almacena información en pares de clave y valor.

**Ejemplo práctico:**

```python
persona = dict(nombre="Ana", edad=22)
print(persona)
```

## 19. Función `enumerate()`

Permite recorrer una colección obteniendo al mismo tiempo el índice y el elemento.

**Ejemplo práctico:**

```python
frutas = ["pera", "uva", "mango"]
for indice, fruta in enumerate(frutas):
    print(indice, fruta)
```

## 20. Función `zip()`

Combina elementos de dos o más iterables posición por posición.

**Ejemplo práctico:**

```python
nombres = ["Ana", "Luis"]
edades = [22, 25]
for nombre, edad in zip(nombres, edades):
    print(nombre, edad)
```

## 21. Función `any()`

Devuelve `True` si al menos un elemento de un iterable es verdadero.

**Ejemplo práctico:**

```python
respuestas = [False, True, False]
print(any(respuestas))
```

## 22. Función `all()`

Devuelve `True` si todos los elementos de un iterable son verdaderos.

**Ejemplo práctico:**

```python
respuestas = [True, True, True]
print(all(respuestas))
```

## 23. Función `pow()`

Calcula una potencia: eleva un número a otro.

**Ejemplo práctico:**

```python
print(pow(2, 3))
```

## 24. Función `divmod()`

Devuelve una tupla con el cociente y el residuo de una división entera.

**Ejemplo práctico:**

```python
print(divmod(17, 5))
```

## 25. Función `ord()`

Devuelve el código Unicode correspondiente a un carácter.

**Ejemplo práctico:**

```python
letra = "A"
print(ord(letra))
```

## 26. Función `chr()`

Devuelve el carácter Unicode correspondiente a un número.

**Ejemplo práctico:**

```python
numero = 65
print(chr(numero))
```

## 27. Función `input()`

Solicita al usuario que escriba un dato. El valor recibido se guarda inicialmente como texto.

**Ejemplo práctico:**

```python
nombre = input("Escribe tu nombre: ")
print("Hola, " + nombre)
```

## 28. Función `map()`

Aplica una función a cada elemento de un iterable. En este ejemplo, convierte cada número en su cuadrado.

**Ejemplo práctico:**

```python
numeros = [1, 2, 3, 4]
cuadrados = list(map(lambda x: x ** 2, numeros))
print(cuadrados)
```

## 29. Función `filter()`

Selecciona los elementos de un iterable que cumplen una condición.

**Ejemplo práctico:**

```python
numeros = [1, 2, 3, 4, 5, 6]
pares = list(filter(lambda x: x % 2 == 0, numeros))
print(pares)
```

## 30. Función `help()`

Muestra la documentación y ayuda de una función, clase o módulo de Python.

**Ejemplo práctico:**

```python
help(len)
```

## Conclusión

Las funciones incorporadas de Python permiten realizar tareas frecuentes como mostrar información, convertir tipos de datos, calcular valores, organizar colecciones y solicitar datos al usuario. Conocerlas facilita la escritura de programas claros y eficientes.
