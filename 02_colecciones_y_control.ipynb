## 1. `range()`

La función `range()` genera una secuencia de números enteros. Puede recibir hasta tres argumentos: inicio, fin y paso. El valor final **no se incluye**.

**Sintaxis:** `range(inicio, fin, paso)`

```python
# Ejemplo 1: secuencia del 1 al 10
numeros = list(range(1, 11))
print("Secuencia generada:", numeros)

# Ejemplo 2: números pares del 2 al 10
pares = list(range(2, 11, 2))
print("Números pares:", pares)
```

## 2. `sum()`

La función `sum()` suma todos los elementos numéricos de un iterable, como una lista o una tupla.

**Sintaxis:** `sum(iterable)`

```python
# Ejemplo: suma de calificaciones
calificaciones = [4.5, 3.8, 5.0, 4.2]
total = sum(calificaciones)

print("Calificaciones:", calificaciones)
print("Suma total:", total)
```

## 3. `min()`

La función `min()` devuelve el elemento con el valor más bajo de una colección.

**Sintaxis:** `min(iterable)`

```python
# Ejemplo: temperatura mínima
temperaturas = [18.5, 22.3, 15.8, 20.1]
temperatura_minima = min(temperaturas)

print("Temperaturas:", temperaturas)
print("Temperatura mínima:", temperatura_minima)
```

## 4. `max()`

La función `max()` devuelve el elemento con el valor más alto de una colección.

**Sintaxis:** `max(iterable)`

```python
# Ejemplo: venta máxima
ventas = [1200, 3450, 2780, 4100]
venta_maxima = max(ventas)

print("Ventas:", ventas)
print("Venta máxima:", venta_maxima)
```

## 5. `sorted()`

La función `sorted()` devuelve una **nueva lista ordenada** a partir de un iterable. No modifica la colección original.

**Sintaxis:** `sorted(iterable, reverse=False)`

```python
# Ejemplo 1: orden alfabético
nombres = ["Carlos", "Ana", "Beatriz", "Daniel"]
nombres_ordenados = sorted(nombres)

print("Lista original:", nombres)
print("Lista ordenada:", nombres_ordenados)

# Ejemplo 2: orden descendente
numeros = [7, 2, 9, 4]
print("Orden descendente:", sorted(numeros, reverse=True))
```

## 6. `reversed()`

La función `reversed()` invierte el orden de los elementos de un iterable. Devuelve un iterador, por lo que generalmente se convierte en lista con `list()`.

**Sintaxis:** `reversed(iterable)`

```python
# Ejemplo: invertir una lista de ciudades
ciudades = ["Bogotá", "Medellín", "Cali", "Barranquilla"]
ciudades_invertidas = list(reversed(ciudades))

print("Orden original:", ciudades)
print("Orden invertido:", ciudades_invertidas)
```

## 7. `enumerate()`

La función `enumerate()` permite recorrer una colección obteniendo al mismo tiempo el índice y el valor de cada elemento.

**Sintaxis:** `enumerate(iterable, start=0)`

```python
# Ejemplo: lista de estudiantes con posición
estudiantes = ["Laura", "Andrés", "Sofía"]

for indice, estudiante in enumerate(estudiantes, start=1):
    print(f"Estudiante {indice}: {estudiante}")
```

## 8. `zip()`

La función `zip()` une dos o más colecciones elemento por elemento. Es útil para recorrer varias listas al mismo tiempo.

**Sintaxis:** `zip(colección1, colección2, ...)`

```python
# Ejemplo: combinar nombres, edades y ciudades
nombres = ["Ana", "Luis", "Marta"]
edades = [20, 25, 22]
ciudades = ["Bogotá", "Cali", "Medellín"]

for nombre, edad, ciudad in zip(nombres, edades, ciudades):
    print(f"{nombre} tiene {edad} años y vive en {ciudad}.")
```

## 9. `abs()`

La función `abs()` devuelve el valor absoluto de un número, es decir, su distancia respecto a cero.

**Sintaxis:** `abs(numero)`

```python
# Ejemplo: valores absolutos
diferencias = [-15, 8, -3.7, 0]
valores_absolutos = [abs(numero) for numero in diferencias]

print("Valores originales:", diferencias)
print("Valores absolutos:", valores_absolutos)
```

## 10. `round()`

La función `round()` redondea un número a la cantidad de decimales indicada. Si no se indica la cantidad de decimales, redondea al entero más cercano.

**Sintaxis:** `round(numero, decimales)`

```python
# Ejemplo: redondeo de un precio
precio = 19.98765

print("Redondeo a 0 decimales:", round(precio))
print("Redondeo a 2 decimales:", round(precio, 2))
print("Redondeo a 3 decimales:", round(precio, 3))
```

## Ejemplo integrador

El siguiente ejemplo combina varias funciones documentadas para realizar un análisis básico de calificaciones.

```python
# Análisis básico de calificaciones
calificaciones = [4.5, 3.2, 4.8, 2.9, 5.0, 3.7]

print("Calificaciones:", calificaciones)
print("Cantidad:", len(calificaciones))
print("Promedio:", round(sum(calificaciones) / len(calificaciones), 2))
print("Nota mínima:", min(calificaciones))
print("Nota máxima:", max(calificaciones))
print("Notas ordenadas:", sorted(calificaciones))
print("Notas en orden inverso:", list(reversed(sorted(calificaciones))))

for posicion, nota in enumerate(calificaciones, start=1):
    print(f"Estudiante {posicion}: {nota}")
```

## Conclusión

Las funciones de colecciones y control permiten procesar datos de forma sencilla y eficiente en Python. Funciones como `sum()`, `min()`, `max()` y `sorted()` facilitan el análisis de listas, mientras que `enumerate()` y `zip()` permiten recorrer colecciones de manera más clara y flexible.
